# Worked example — physics quiz with rich content

This source notebook combines every feature the package supports in one
place, framed as a short undergraduate physics problem set:

- **Single-choice (SC)** and **many-choice (MC)** with MathJax in the
  question, answer labels, and per-answer feedback.
- **Numeric (NM)** with both exact-value and range matchers, with and
  without precision; MathJax in feedback strings.
- A **code-block question** showing how to embed Python source as the
  question prompt.
- **Per-question points** via `{N}`, including fractional weights.
- A **self-check quiz inside a graded task cell** (`graded=false`) so
  the instructor can read the prose, grade it manually, and let the
  student warm up with the quiz first.

Each task cell below is annotated with *what's being demonstrated* and
*why that combination is interesting* — so the notebook reads as a
tour, not just a fixture.

Open this notebook in a course directory under
`source/<assignment>/`, mark each task cell as *Manually Graded Task*
in the nbgrader cell toolbar (the metadata is already pre-populated
below — JupyterLab should pick it up), and run
`nbgrader generate_assignment` to produce a release.

## Warm-up — self-check inside a graded task (`warmup`)

*What's being demonstrated:* `graded=false` opts a quiz out of
auto-grading.  The host cell stays a Manually Graded Task so the
instructor can read and score the prose around it; the quiz itself is
purely formative — the student gets immediate green / red feedback.

Before answering the graded questions below, sanity-check your unit
intuition:

#### Quiz encoded=false graded=false
* (SC) "Which of these is a unit of energy?"
  + "Joule (J)"
  - "Newton (N)"        (That's a unit of force.)
  - "Pascal (Pa)"       (That's a unit of pressure.)
  - "Watt (W)"          (That's a unit of power.)
#### End Quiz

## MathJax in question and answers (`kinematics-mc`)

*What's being demonstrated:* MathJax (`$...$`) in question and
answer label.  Per-question points via `{N}`, here unevenly
distributed across two questions.  Default graded mode
(`hide_correctness` auto-on) shows neutral *Selected: ...*
feedback rather than green / red.

*Why this combination matters:* graded MathJax content is the most
common content category in STEM courses; this is the realistic
case.

*A note on per-answer feedback:* it's deliberately omitted here.
Graded mode hides correctness colours, but per-answer feedback
strings still ship to the student's browser and would reveal which
answer is which.  See *Where the answer key lives* in
`graded-quizzes` for what the redaction does and doesn't cover.

#### Quiz
* (MC) {3} "Which of these are equivalent expressions for kinetic energy?"
  + "$\tfrac{1}{2} m v^2$"
  + "$\tfrac{p^2}{2m}$"
  - "$m v$"
  - "$m g h$"
* (SC) {2} "At what speed does relativistic KE exceed Newtonian by 10%?"
  + "$\approx 0.4 c$"
  - "$\approx 0.1 c$"
  - "$\approx 0.9 c$"
#### End Quiz

## Numeric: exact value, range, precision (`numeric-energies`)

*What's being demonstrated:* both forms of numeric answer.  The
first question uses an exact-value match `+ <v>` with a `[3]`
precision marker — the student's input is rounded to 3
significant figures before comparison.  The second uses a range
`+ [min, max]` with tolerance bands `- [min, max]` (incorrect)
that the autograder will use to score, but no per-answer
feedback (those would leak the correct band by elimination — see
the kinematics cell note).  A bare `- (Hint: ...)` line is the
catch-all default; in numeric questions the JS overrides
per-answer feedback in hide mode anyway, so the hint is neutral
formula context, not a correctness signal.

*Why this combination matters:* physics quantities don't have a
canonical decimal expansion.  Precision-rounded value matches
handle the "3 sig. figs." case; ranges handle anything where the
answer depends on which constant table the student looked up.

#### Quiz
* (NM) {2} "Speed of light in m/s, 3 sig. figs." [3]
  + <3.00e8>
  - (Hint: $c$ is around $3 \times 10^8$ m/s.)
* (NM) {3} "Energy of a photon at $\lambda = 500$ nm, in eV."
  + [2.4, 2.6]
  - [1.0, 2.3]
  - [2.7, 5.0]
  - (Hint: $E = hc / \lambda$ with $hc \approx 1240$ eV·nm.)
#### End Quiz

## Code-block question (`code-block-question`)

*What's being demonstrated:* the question line carries both prose and
a Python code block (\`\`\`...\`\`\`).  Within a code block the
delimiters that normally bound text — `\"`, `(`, `)` — pass through
verbatim, which is the workaround for the `\"`/`)` parser limitation
(see *Parser limitations* in the quiz-syntax docs).  Use literal
`\n` for line breaks inside the code block; the parser substitutes
real newlines.

*Why this combination matters:* showing source code as the prompt is
common in computational-physics and programming-adjacent material; it
also exercises the only parser-safe way to embed text containing
double quotes.

#### Quiz
* (SC) {1.5} "What does this Python function compute?" ```def f(v, m):\n    return 0.5 * m * v**2```
  + "Newtonian kinetic energy of a particle"
  - "Relativistic kinetic energy"
  - "Momentum of a particle"
#### End Quiz

## Try it

From a course directory whose `nbgrader_config.py` registers
`CreateQuiz` (see the [tutorial](../tutorial.html)):

```
cp nb3-physics-rich-content.ipynb source/ps1/
nbgrader generate_assignment ps1 --force
```

The release notebook in `release/ps1/` shows interactive widgets in
place of each `#### Quiz` block.  Open one, click some answers,
submit a numeric value or two; you'll see a `responses.json` file
appear in the same directory.  Run `nbgrader autograde` and the
graded cells are scored against the embedded answer key with
partial-credit support.

Total points across the four task cells of this notebook:
`3 + 2 + 2 + 3 + 1.5 = 11.5` (the warm-up cell is unweighted).
